# Health Data Analytics — Starter Notebook

This notebook loads the seven HDA datasets and shows a few basic explorations to get you going. See `hda_data.py` for the loader and `README.md` for setup.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import hda_data as hda

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

## Overview of all datasets

```python
hda.summary()
```

## Load everything

`load_all()` returns a dict keyed by the short names shown above (`ehr`, `operations`, `satisfaction`, `billing`, `clinical_trials`, `iot_wearable`, `public_health`).

```python
data = hda.load_all()
ehr = data["ehr"]
ehr.head()
```

## Example 1: EHR patient age distribution

```python
ax = ehr["Age"].plot(kind="hist", bins=30, edgecolor="white")
ax.set_title("EHR — Patient Age Distribution")
ax.set_xlabel("Age")
plt.show()
```

## Example 2: Patient satisfaction ratings

```python
sat = data["satisfaction"]
ax = sat["Satisfaction_Rating"].value_counts().sort_index().plot(kind="bar")
ax.set_title("Patient Satisfaction Ratings")
ax.set_xlabel("Rating")
ax.set_ylabel("Count")
plt.show()
```
